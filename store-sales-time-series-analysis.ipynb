# **Step 1 — Import Libraries**

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

# Ignore warnings
import warnings
warnings.filterwarnings("ignore")

print("Libraries imported successfully.")
```

# **Step 2 — Load Dataset**

```python
df = pd.read_csv(
    "/kaggle/input/datasets/muhammadumer7804/store-sales-time-series-forecasting-dataset/stores_sales_forecasting.csv",
    encoding="latin1"
)

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

# **Step 3 — Basic Exploration**

```python
print("=" * 60)
print("First 5 Rows")
print("=" * 60)

display(df.head())
```

# **Step 4 — Dataset Information**

```python
print("=" * 60)
print("Dataset Information")
print("=" * 60)

df.info()
```

# **Step 5 — Data Cleaning**

```python
# Convert date columns
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

print("Date columns converted successfully.\n")
```

```python
# Check missing values
print("Missing Values")
print(df.isnull().sum())

print("\nDuplicate Rows :", df.duplicated().sum())
```

# **Step 6 — Feature Engineering (Time Series)**

```python
# Create time-based features
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter

print("Feature Engineering completed successfully.\n")

print(df[["Order Date", "Year", "Month", "Month Name", "Quarter"]].head())
```

# **Step 7 — Monthly Sales Trend (Most Important Chart)**

```python
# Monthly Sales Trend


# Monthly Sales
monthly_sales = (
    df.groupby(pd.Grouper(key="Order Date", freq="M"))["Sales"]
      .sum()
)

# Plot
plt.figure(figsize=(12,5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    linewidth=2
)

plt.title("Monthly Sales Trend")
plt.xlabel("Order Date")
plt.ylabel("Total Sales")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

# **Step 8 — 3-Month Rolling Average**

```python

# Month Rolling Average


# Calculate 3-Month Rolling Average
rolling_avg = monthly_sales.rolling(window=3).mean()

plt.figure(figsize=(12,5))

# Original Sales
plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    label="Monthly Sales"
)

# Rolling Average
plt.plot(
    rolling_avg.index,
    rolling_avg.values,
    linewidth=3,
    label="3-Month Rolling Average"
)

plt.title("Monthly Sales vs Rolling Average")
plt.xlabel("Order Date")
plt.ylabel("Sales")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

# **Step 9 — Year-wise Sales Comparison**

```python

#  Year-wise Sales


yearly_sales = (
    df.groupby("Year")["Sales"]
      .sum()
)

plt.figure(figsize=(8,5))

yearly_sales.plot(kind="bar")

plt.title("Year-wise Sales")
plt.xlabel("Year")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

# **Step 10 — Quarterly Sales Trend (Final Chart)**

```python

#Quarterly Sales Trend


quarterly_sales = (
    df.groupby("Quarter")["Sales"]
      .sum()
)

plt.figure(figsize=(8,5))

quarterly_sales.plot(
    kind="bar"
)

plt.title("Quarter-wise Sales")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
